```python
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("MahmoudIbrahim/Meta-LLama3-Instruct-Arabic")
```

```python
txt = "السلام عليكم"
tokenizer.tokenize(txt)
```

The previous output `['Ø§ÙĦ', 'Ø³ÙĦØ§Ùħ', 'ĠØ¹ÙĦÙĬÙĥ', 'Ùħ']` is a common display issue where UTF-8 characters are misinterpreted. However, the tokens themselves are likely correct. We can verify this by decoding them back into a human-readable string.

```python
token_ids = tokenizer.encode(txt)
decoded_text = tokenizer.decode(token_ids)
print(f"Original text: {txt}")
print(f"Token IDs: {token_ids}")
print(f"Decoded text from tokens: {decoded_text}")
```
