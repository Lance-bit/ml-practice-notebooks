```python
import re
import string
```

```python
raw_text = """
<div>
  OMG!!! The movie was SOOOO bad 😡.
  Check the review at https://Movies.com/Bad-Review
  #disappointed
</div>
"""
```

```python
def HTMLclean_text(text):
  html_pattern = r'<.*?>'

  step1 = re.sub(html_pattern,'',text)
  return step1
```

```python
raw_text = HTMLclean_text(raw_text)
raw_text
```

```python
def removeupper(text):
  step2 = text.lower()

  return step2
```

```python
raw_text = removeupper(raw_text)
```

```python
print(raw_text)
```

```python
def replaceemojis(text):
  list = ["😡","😀","😃","😁","😂"]

  for i in text:
    if i in list:
      text = text.replace(i,"")

  return text
```

```python
raw_text = replaceemojis(raw_text)
```

```python
def removeurl(text):
  url_pattern = r'https?://\S+|www\.\S+'
  text =re.sub(url_pattern,'',text)

  return text
```

```python
raw_text=removeurl(raw_text)
```

```python
print(raw_text)
```

```python
def removesymbols(text):
  symbols_pattern = r'[^a-zA-Z0-9\s]'
  text = re.sub(symbols_pattern,'',text)

  return text
```

```python
raw_text = removesymbols(raw_text)
```

```python
print(raw_text)
```
